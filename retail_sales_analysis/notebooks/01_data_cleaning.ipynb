# 01 · Data Loading & Cleaning

**Goal:** Load the raw Superstore CSV, understand its structure, clean it, and save a cleaned version for downstream notebooks.

---
### Steps
1. Load raw data
2. Inspect shape, dtypes, sample rows
3. Check for nulls and duplicates
4. Fix data types (dates, numerics)
5. Feature engineering (derived columns)
6. Save cleaned dataset

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_superstore, clean_data, feature_engineer, quick_summary

%matplotlib inline
pd.set_option('display.max_columns', 50)
pd.set_option('display.float_format', '{:.2f}'.format)
```

## 1. Load Raw Data

```python
df_raw = load_superstore()
print(f'Shape: {df_raw.shape}')
df_raw.head()
```

## 2. Inspect Structure

```python
print('--- Data Types ---')
print(df_raw.dtypes)
print('\n--- Basic Stats ---')
df_raw.describe(include='all')
```

## 3. Check Nulls & Duplicates

```python
print('Null counts:')
print(df_raw.isnull().sum())
print(f'\nDuplicate rows: {df_raw.duplicated().sum()}')
```

## 4. Clean Data

```python
df_clean = clean_data(df_raw)
df_clean.dtypes
```

## 5. Feature Engineering

```python
df = feature_engineer(df_clean)
df[['Order Date', 'Ship Date', 'Order Year', 'Order Month',
    'Order Quarter', 'Order YearMonth', 'Ship Lag (days)',
    'Sales', 'Profit', 'Profit Margin (%)']].head()
```

## 6. Quick Business Summary

```python
quick_summary(df)
```

## 7. Save Cleaned Data

```python
output_path = '../data/superstore_clean.csv'
df.to_csv(output_path, index=False)
print(f'Saved cleaned dataset → {output_path}')
```
